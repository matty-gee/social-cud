### other behavioral signatures?

could be something in RT... theres a trend towards them taking longer on affiliation trials than power trials, maybe we need a model that regresses out trial-level structure to clean it up some... maybe regress out character-specific variations etc...

```python
def rt_diffs(rts, eps=1e-6):
    rt_mask = np.isfinite(rts) & (rts > 0)
    rt = np.log(rts[rt_mask] + eps)
    dim = np.asarray(decision_trials["dimension"])[rt_mask]

    aff = rt[dim == "affil"]
    pow = rt[dim == "power"]
    if aff.size == 0 or pow.size == 0:
        return np.nan
    return float(np.mean(aff) - np.mean(pow))

def social_dist(dists):
    dists_by_char = organize_by_character(dists)
    dists_by_char = [dists[-1] for dists in dists_by_char]
    return np.mean(dists_by_char)

features = {"rt_diffs": lambda df: rt_diffs(df['reaction_time'].to_numpy()), 
            "social_distance": lambda df: social_dist(df['distance'].to_numpy())}
df_sub = add_summary_behavior(incl_subs, features)
df_sub = df_sub.merge(data, on='sub_id')
df_sub_cd = df_sub[df_sub['dx'] == 1]
print_df(run_regression('rt_diffs ~ coc_age_1st_use + coc_dur_years + reaction_time_mean + missing_trials + age_years', data=df_sub_cd))
```
