```python
#!/usr/bin/env python3
"""
Copy each <SUBID>_beta_4d.nii from SRC into:
  DST/sub-<SUBID>/beta_decisions.nii.gz

- Does NOT modify originals.
- Gzips on the fly (so output is .nii.gz).
"""

from __future__ import annotations

import gzip
import re
import shutil
from pathlib import Path

SRC = Path("/Users/matty_gee/Desktop/SocialTrajectories/analyses/lsa/imgs")
DST = Path("/Users/matty_gee/Desktop/SocialTrajectories/analyses/lsa/subs")

PAT = re.compile(r"^(?P<sid>\d+)_beta_4d\.nii$", re.IGNORECASE)


def copy_and_gzip(src_nii: Path, dst_niigz: Path, *, overwrite: bool = True) -> None:
    dst_niigz.parent.mkdir(parents=True, exist_ok=True)

    if dst_niigz.exists() and not overwrite:
        print(f"SKIP (exists): {dst_niigz}")
        return

    with src_nii.open("rb") as fin, gzip.open(dst_niigz, "wb", compresslevel=6) as fout:
        shutil.copyfileobj(fin, fout)

    print(f"WROTE: {dst_niigz}")


def main() -> None:
    files = sorted(SRC.glob("*_beta_4d.nii"))
    if not files:
        raise SystemExit(f"No files matching '*_beta_4d.nii' found in: {SRC}")

    for f in files:
        m = PAT.match(f.name)
        if not m:
            print(f"SKIP (name mismatch): {f.name}")
            continue

        sid = m.group("sid")
        out = DST / f"sub-{sid}" / "beta_decisions.nii.gz"
        copy_and_gzip(f, out, overwrite=True)


if __name__ == "__main__":
    main()
```
