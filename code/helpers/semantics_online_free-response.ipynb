```python
def _clean_text_list(xs):
    """Return list[str] with NaN/None -> '' and whitespace normalized."""
    out = []
    for x in xs:
        if x is None or pd.isna(x):
            out.append("")
        else:
            out.append(" ".join(str(x).split()))
    return out

llm = "openai"
out_dir = "../data/other-samples/online/narratives/free-responses"
os.makedirs(out_dir, exist_ok=True)

cols = [f"free_response_{c}" for c in CHARACTERS]

for sub_id in incl_subs_online:
    out_fname = f"{out_dir}/sub-{sub_id}_free-responses_{llm}.npz"
    if os.path.exists(out_fname):
        print(f"Skipping sub-{sub_id}, already done.")
        continue

    sub_rows = data_online.loc[data_online["sub_id"] == sub_id]

    # Skip if subject not present
    if sub_rows.shape[0] == 0:
        print(f"Skipping sub-{sub_id}: not found in data_online.")
        continue

    # Skip if any expected columns are missing from the dataframe
    missing_cols = [c for c in cols if c not in sub_rows.columns]
    if missing_cols:
        print(f"Skipping sub-{sub_id}: missing columns {missing_cols}.")
        continue

    # Take first row (if duplicates exist)
    raw = sub_rows.loc[:, cols].iloc[0].tolist()
    free_resps = _clean_text_list(raw)

    # Skip if they have no free responses at all (all empty after cleaning)
    if all(s == "" for s in free_resps):
        print(f"Skipping sub-{sub_id}: no free responses.")
        continue

    semantic_vectors = get_sentence_embeddings(free_resps, model=llm, normalize=True)

    np.savez_compressed(
        out_fname,
        text=np.array(free_resps, dtype=object),
        embedding=np.asarray(semantic_vectors),
        character_order=np.array(CHARACTERS, dtype=object),
    )
```
